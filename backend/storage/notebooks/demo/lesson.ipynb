## 什么是变量？

变量就像一个**带标签的盒子**，你可以把数据放进去，然后通过标签来访问它。



> 💡 **提示**：Python 中变量不需要声明类型，解释器会自动推断。


```python
# 试试修改变量的值
name = "Alice"
age = 20
height = 1.75
is_student = True

print("姓名:", name)
print("年龄:", age)
print("身高:", height)
print("是否是学生:", is_student)
```

## 字符串操作

字符串可以**拼接**（+）和**重复**（*）：



还可以用  把变量嵌入字符串中。


```python
first_name = "张"
last_name = "三"
full_name = first_name + last_name
print("全名:", full_name)

# f-string 格式化
age = 20
print(f"我叫{full_name}，今年{age}岁")

# 试试修改下面的代码
greeting = "你好"
print(greeting * 3)
```
